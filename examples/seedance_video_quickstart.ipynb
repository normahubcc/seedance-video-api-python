## 2. Инициализация клиента
Ключ выпускается в личном кабинете [normahub.cc](https://normahub.cc). Оплата в рублях (карты РФ, СБП).

```python
import os
from seedance import NormaVideoClient

API_KEY = os.getenv("NORMAHUB_API_KEY", "nr_hub_your_key_here")
client = NormaVideoClient(api_key=API_KEY)
print("✓ Клиент NormaHub SeeDance успешно инициализирован!")
```

## 3. Генерация видео из текста (Text-to-Video)
Создаём задачу генерации видео высокого разрешения с моделью `seedance-2.5`.

```python
prompt = "Киберпанк мегаполис, неоновый дождь, парящие дроны, кинематографичная камера 4K"

print("Отправка задачи на генерацию...")
task = client.create_task(
    prompt=prompt,
    model="seedance-2.5",
    duration=5,
    aspect_ratio="16:9"
)
print(f"Задача создана! ID задачи: {task.id}")
```

## 4. Ожидание готовности (Polling)
Метод `wait_for_task` опрашивает статус до завершения рендера нейросетью.

```python
print("Рендеринг видео... Ожидаем...")
completed_task = client.wait_for_task(task.id, timeout=300, poll_interval=4)
print(f"Статус: {completed_task.status}")
print(f"URL готового MP4: {completed_task.video_url}")
```

## 5. Встроенный плеер в блокноте

```python
from IPython.display import HTML

if completed_task.video_url:
    display(HTML(f"""
    <video width="640" height="360" controls autoplay muted>
      <source src="{completed_task.video_url}" type="video/mp4">
      Браузер не поддерживает HTML5 видео.
    </video>
    """))
```
